```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from keras.models import Sequential
from keras.layers import Dense
from tensorflow.keras.optimizers import Adam

# 1. LOAD THE DATA
df = pd.read_csv('train_data.csv')

# --- FEATURE ENGINEERING (MÜHENDİSLİK DOKUNUŞU) ---
# AI'a kendi genişliğini hesaba katması için aradaki net mesafeyi veriyoruz!
df['Delta_X'] = df['Player_X'] - df['Closest_Rock_X']

# --- THE ANTI-LAZY AI TRICK ---
stay_data = df[df['Action_Taken'] == 1]
move_data = df[df['Action_Taken'] != 1]

# Filtreyi %35'te tutuyoruz ki hem kaçsın hem de gereksiz yere sağa sola gitmesin
if len(stay_data) > len(move_data):
    stay_data = stay_data.sample(frac=0.35, random_state=42)

df = pd.concat([stay_data, move_data]).sample(frac=1).reset_index(drop=True)

# 2. SPLIT FEATURES AND TARGET
# DİKKAT: Artık 3 değil, 4 kolon veriyoruz (Delta_X eklendi)
X = df[['Player_X', 'Closest_Rock_X', 'Closest_Rock_Y', 'Delta_X']]
y = df['Action_Taken']

# 3. NORMALIZE THE DATA
X_scaled = X / 800.0

# 4. SPLIT
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 5. BUILD THE NEURAL NETWORK 
model = Sequential()
# DİKKAT: input_dim=4 oldu çünkü artık 4 bilgi alıyor!
model.add(Dense(64, activation='relu', input_dim=4))
model.add(Dense(32, activation='relu'))
model.add(Dense(3, activation='softmax')) 

# 6. COMPILE & TRAIN
model.compile(loss='sparse_categorical_crossentropy', optimizer=Adam(learning_rate=0.005), metrics=['accuracy'])
print("Feature Engineering uygulanmış model eğitiliyor...")
model.fit(X_train, y_train, epochs=50, batch_size=32, validation_data=(X_test, y_test), verbose=1)

# 7. SAVE
model.save('ai_player.keras')
print("✅ Yeni ve Çok Daha Zeki Beyin Hazır!")
```
