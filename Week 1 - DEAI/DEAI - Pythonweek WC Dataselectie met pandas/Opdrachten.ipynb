Geef een overzicht met daarin van de producten behorend tot het producttype ‘Cooking Gear’ per productnaam de totaal verkochte hoeveelheid en de gemiddelde verkoopprijs. Sorteer de uitkomst op totaal verkochte hoeveelheid. (6 kolommen, 10 rijen)

Geef een overzicht met daarin de voornaam en de achternaam van de medewerkers die nog nooit wat hebben verkocht (2 kolommen, 25 rijen)

Geef een overzicht met daarin het aantal producten waarvan de marge lager is dan de gemiddelde marge van alle producten samen. Geef in het overzicht tevens aan wat de gemiddelde marge is van dit aantal producten waarvan de marge lager dan de gemiddelde marge van alle producten samen is. (1 kolom, 2 rijen) 

```python
# 1. Bereken de gemiddelde marge van ALLE producten
gemiddelde_marge_totaal = product_df['MARGIN'].mean()

# 2. Filter producten met een marge LAGER dan dit globale gemiddelde
lage_marge_producten = product_df.loc[product_df['MARGIN'] < gemiddelde_marge_totaal, :]

# 3. Bereken hoeveel dit er zijn, en wat hún specifieke gemiddelde is
aantal_producten = lage_marge_producten['PRODUCT_NUMBER'].count()
gemiddelde_marge_specifiek = lage_marge_producten['MARGIN'].mean()

# 4. Stop dit in een net overzichtje (1 kolom, 2 rijen)
resultaat_marge = pd.DataFrame(
    [aantal_producten, gemiddelde_marge_specifiek],
    columns=['Waarde'],
    index=['Aantal producten', 'Gemiddelde marge van deze groep']
)

print(resultaat_marge)
```

Geef een overzicht met daarin de namen van de producten die voor meer dan 500 (verkoopprijs) zijn verkocht maar nooit zijn teruggebracht. (1 kolom, 13 rijen) 

```python
# 1. Filter order_details op verkoopprijs hoger dan 500
dure_orders = order_details_df.loc[order_details_df['UNIT_SALE_PRICE'] > 500, :]

# 2. Maak een lijstje (subselect) van alle codes die geretourneerd zijn
geretourneerd_lijst = returned_item_df['ORDER_DETAIL_CODE'].drop_duplicates()

# 3. Filter dure orders: behoud alleen degene die NIET (~) geretourneerd zijn
nooit_terug_orders = dure_orders.loc[~dure_orders['ORDER_DETAIL_CODE'].isin(geretourneerd_lijst), :]

# 4. Merge met product_df om de naam van het product op te halen
merged_producten = pd.merge(nooit_terug_orders, product_df, on='PRODUCT_NUMBER', how='inner')

# 5. Selecteer de productnamen en verwijder dubbelingen (zodat je 13 unieke rijen krijgt)
resultaat_producten = merged_producten[['PRODUCT_NAME']].drop_duplicates()

print(resultaat_producten)
```

Geef met behulp van bovenstaande hulpcode een overzicht met daarin op basis van het aantal jaar dat iemand in dienst is of een medewerker ‘kort in dienst’ (minder dan 30 jaar in dienst) of een ‘lang in dienst’ (groter dan gelijk aan 30 jaar in dienst) is. Geef daarbij per medewerker in een aparte kolom zowel ‘kort in dienst’ als ‘lang in dienst’ aan. Gebruik (wederom) een for-loop.<br>
(2 kolommen, 102 rijen) 

```python
from datetime import datetime, date

# 1. Haal het huidige jaar op (via de methode van de docent)
huidig_jaar = date.today().year

# Kopieer de kolommen die we nodig hebben
dienst_df = sales_staff_df[['LAST_NAME', 'DATE_HIRED']].copy()

# 2. Loop door de medewerkers heen
for index, row in dienst_df.iterrows():
    date_str = str(row['DATE_HIRED'])
    
    # Soms staan er tijdstippen achter de datum (zoals 00:00:00). We pakken de eerste 10 tekens.
    date_str_short = date_str[:10]
    
    # Omdat SQLite datums zowel in 'DD-MM-YYYY' als 'YYYY-MM-DD' opslaat,
    # doen we dit veilig via een 'try-except' om fouten te voorkomen:
    try:
        # Probeer het format van de hint van de docent
        date_obj = datetime.strptime(date_str_short, '%d-%m-%Y')
    except ValueError:
        # Werkt dat niet, gebruik dan deze (standaard voor databases)
        date_obj = datetime.strptime(date_str_short, '%Y-%m-%d')
        
    # 3. Reken de dienstjaren uit
    jaar_in_dienst = date_obj.year
    jaren_werkzaam = huidig_jaar - jaar_in_dienst
    
    # 4. Vul de nieuwe kolom in met de for-loop logic
    if jaren_werkzaam < 30:
        dienst_df.at[index, 'Dienstverband'] = 'kort in dienst'
    else:
        dienst_df.at[index, 'Dienstverband'] = 'lang in dienst'

# Print alleen de twee gevraagde kolommen
resultaat_dienst = dienst_df[['LAST_NAME', 'Dienstverband']]

print(resultaat_dienst)
```
